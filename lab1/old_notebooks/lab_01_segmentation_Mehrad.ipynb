### 1.4 Region growing (5pts)

In this exercise you will implement from scratch your own region growing algorithm. Choose one of the image channel (R/G/B/H) as the source image. You can use multiple seed point to achieve better results. Plot the best result along with the error rate. Save the mask as the variable `mask_region`.

##### Comments: 
We chose hue as the main signal because despite its compact histogram, it is rather homogenous in the target areas. <br>
First, we tried implementing the algorithm with a recursive function `region_growing` but for some reason we had a memory issue. So we used queues. Function `region_growing_qu` gets an initial x y coordinates and puts it in a queueu. In each iteration it pops one point out of the queueu, checks the criterion and if the criterion was satisfied it pushes the 4 neighbours of this point in the queueu until there is no point left in the queueu. Matrix M is the mask matrix it also keeps track of the previously checked point to avoid checking them again. Function `region_growing_check` just checks the critetia. <br>
First, we set the criteria to be a simple threshold but it was too similar to the tresholding method so we choose another criteria. the current criteria is to check for sudden changes in hue. We dicard points that have a high hue difference from their neighbour. By manually optimizing this threshold we can see that 0.1 is the best value. <br>
Instead of growing the target region we grow the background because it is one piece and the segmentation can be done by calling the function only once. We choose the initial point by visually picking a background point.  <br>
At the end we open the mask with the parameter 400 to get rid of tiny defects this reduced error by 2 percent. The final error is 12.7 percent


```python
def region_growing(x,y,P,M,th, h1,h2):
    if x >= h1 or x<0 or y>=h1 or y<0: 
        return M 
    if M[x,y] != 0:
        return M 
    elif P[x,y] > th:
        M[x,y] = -1 
        return M 
    else: 
        M[x,y] = 1  
        M = region_growing(x+1,y+1,P,M,th,h1,h2)
        M = region_growing(x+1,y-1,P,M,th,h1,h2)  
        M = region_growing(x-1,y+1,P,M,th,h1,h2)
        M = region_growing(x-1,y-1,P,M,th,h1,h2)
        return M 
        


```

```python
def region_growing_check(x,y,P,M,th,b,h1,h2):
    if x<0 or x>=h1 or y<0 or y>=h2:
        return 0,M 
    if M[x,y] != 0: 
        return 0,M 
    else: 
        if np.abs(P[x,y]-b) > th:
            M[x,y] = -1 
            return 0,M
        else: 
            M[x,y] = 1 
            return 1,M
```

```python
from collections import deque
def region_growing_qu(x,y,P,M,th,b):
    sh = np.shape(P)
    h1 = sh[0]
    h2 = sh[1] 
    q = deque([(x,y,th)])
    while len(q) > 0:
        f = q.pop()
        i = f[0]
        j = f[1]
        th = f[2]
        t,M = region_growing_check(i,j,P,M,th,b,h1,h2) 
        if t ==1: 
            temp = P[i,j]
            q.appendleft((i+1,j,th,temp))
            q.appendleft((i-1,j,th,temp))
            q.appendleft((i,j+1,th,temp))
            q.appendleft((i,j-1,th,temp))
            
    
    return M
           
```

```python
# TODO
import skimage.morphology as mor
import random
import skimage.color as col
mask_region = np.zeros_like(adeno_gt)
M= np.zeros_like(adeno_gt)
hsv_img = col.rgb2hsv(adeno_img)
h = hsv_img[:, :, 0]
sh = np.shape(h)
fig , ax = plt.subplots(1, 3, figsize=(15, 50))
x= 1000
y= 100
M = region_growing_qu(x,y,h,M , 0.11, h[x,y])
M = 1-M
ax[0].imshow(M)
error  =1- (np.sum(M==adeno_gt)/sh[0]/sh[1])
ax[0].set_title('Mask after region growing\n error:{}'.format(error))

P =  mor.area_opening(M,400)
plt.imshow((P))
error  = 1- (np.sum(P==adeno_gt)/sh[0]/sh[1])
ax[1].imshow(P)
ax[1].set_title('Final Mask after region growing and opening\n error:{}'.format(error))
ax[2].imshow(adeno_gt)
ax[2].set_title('anedo_gt')
mask_region = P
```

#### Intro 

Axon segmentation is vital to neuroscience research as it enables the study of neurons' morphology and connectivity. We will conduct two exercises to evaluate the impact of your segmentation algorithms under different independent artefacts. Each exercise consists of a synthetic **Ground Truth (GT)** and two more images simulating a specific artifact.

The **Ground Truth**  has tree main compartments. 1) The inner part of the axon. 2) The outer layer of the axon (called myelin) and the extracellular slace (the background of the image). 

The task consists into segmenting the three main compartments in the image to compute the inner-axon and outer-axon radii distribution.


To compute the radii, we suggest the following approximation.
$$\hat{r}_{I}= \sqrt{\frac{A_{yellow}}{\pi}}$$
$$\hat{r}_{O}= \sqrt{\frac{A_{yellow} + A_{blue}}{\pi}}$$
Where **A** is the corresponding area of the comparments. 
**Using this image as reference**
