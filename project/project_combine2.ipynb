In this project, you will be working on solving tiling puzzles using image analysis and pattern recognition techniques. Tiling puzzles are a classic type of puzzle game that consists of fitting together pieces of a given shape (in this case squared to form a complete image. The goal of this project is to develop an algorithm that can automatically reconstruct tiling puzzles from a single input image.

````python
def rotate_image(image, angle, center):
    aa = int(180*angle/np.pi)
    if aa < -45:
        aa = 90+aa
    rot_mat = cv2.getRotationMatrix2D(center, aa , 1.0)
    result = cv2.warpAffine(image, rot_mat, image.shape[1::-1], flags=cv2.INTER_LINEAR)
    return result, rot_mat

def find_next_vertice(v,k):
    li = np.shape(v)[0]
    v0 = v[k,0,:].reshape(2,)
    if k+1>=li:
        return v0,k
    else:
        v0 = v[k,0,:].reshape(2,)
        i = k+1
        vt = v[i,0,:].reshape(2,)
        d= np.linalg.norm(vt-v0)
        while d < 90 and i<li:
            i = i+1
            vt = v[i,0,:].reshape(2,)
            d= np.linalg.norm(vt-v0)
        return vt,i
````

````python
def segment(image , save=False, show=False):
    """
    param image: input image of the puzzle
    param save: if true saves the tiles as jpg files
    param tiles: a list containing all tiles each one 128*128*3 images
    """
    #Preprocessing
    r = image[:, :, 0]
    g = image[:, :, 1]
    b = image[:, :, 2]
    hsv_img = col.rgb2hsv(image)
    h = hsv_img[:, :, 0]
    s = hsv_img[:, :, 1]
    v = hsv_img[:, :, 2]
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    h1 = 255*h/h.max()
    s1 = 255*s/s.max()
    s1= s1.astype(np.uint8)
    h1= h1.astype(np.uint8)
    hue_enhanced =cv2.GaussianBlur(h1,(5,5),5)
    r_enhanced =cv2.bilateralFilter(r,9,75,75)
    g_enhanced =cv2.bilateralFilter(g,9,75,75)
    b_enhanced =cv2.bilateralFilter(b,9,75,75)
    s_enhanced =cv2.bilateralFilter(s1,9,75,75) 
    # Enhancement is done by bilateral filter to preserve very sharp edges
    gray_enhanced = cv2.bilateralFilter(gray,9,75,75)

    # Thresholding
    hue_th = cv2.adaptiveThreshold((hue_enhanced).astype(np.uint8),255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,11,2)
    r_th = cv2.adaptiveThreshold((r_enhanced).astype(np.uint8),255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,11,2)
    g_th = cv2.adaptiveThreshold((g_enhanced).astype(np.uint8),255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,11,2)
    b_th = cv2.adaptiveThreshold((b_enhanced).astype(np.uint8),255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,11,2)
    s_th = cv2.adaptiveThreshold((s_enhanced).astype(np.uint8),255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,11,2)
    gray_th = cv2.adaptiveThreshold((gray_enhanced).astype(np.uint8),255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,11,2)
    th = (np.minimum(gray_th,r_th)).astype(np.uint8) 
    # finding the min of different thresholds to have all the details
    th = (np.minimum(th,g_th)).astype(np.uint8)
    th = (np.minimum(th,b_th)).astype(np.uint8)
    th = (np.minimum(th,s_th)).astype(np.uint8)

    # Morphology
    mask = cv2.morphologyEx(th,cv2.MORPH_CLOSE, np.ones((1,1))) #deleting small noises
    mask = cv2.morphologyEx(mask,cv2.MORPH_OPEN, np.ones((3,3))) #filling inside the tiles
    mask = 255-mask

    # Finding Tiles
    cnts = cv2.findContours((mask), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    cnts = cnts[0] if len(cnts) == 2 else cnts[1]
    margin = 40
    min_area = 10000
    max_area = 20000
    image_number = 0
    ROIS = []
    ROIcnts= []
    ROIms = []
    for c in cnts:
        area = cv2.contourArea(c)
        if area > min_area and area < max_area: # tile detection is based on the area of the contours
            x,y,w,h = cv2.boundingRect(c)
            ROIt = image[max(y-margin,0):min(y+h+margin,2000), max(x-margin,0):min(x+w+margin,2000)]
            ROImt = mask[max(y-margin,0):min(y+h+margin,2000), max(x-margin,0):min(x+w+margin,2000)]
            ROI =  cv2.copyMakeBorder(ROIt, 0, 0, 100, 100,cv2.BORDER_CONSTANT  , None, [0,0,0])
            ROIm = cv2.copyMakeBorder(ROImt, 0, 0, 100, 100,cv2.BORDER_CONSTANT , None, [0,0,0])
            ROIS.append(ROI)
            ROIms.append(ROIm)
            roi_cnts = cv2.findContours((ROIm), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
            roi_cnts = roi_cnts[0] if len(roi_cnts) == 2 else roi_cnts[1]
            for cc in roi_cnts:
                aarea = cv2.contourArea(cc)
                if aarea > min_area and aarea < max_area:
                    ROIcnts.append(cc)
            image_number += 1

    # Rotating each tile
    tiles = []
    for i, roi in enumerate(ROIS):
        c = ROIcnts[i]
        roim = ROIms[i]

        vertices = cv2.approxPolyDP(c, 0.01*cv2.arcLength(c, True), True) # for each ROI we fit a rectangle and get the vertices
        v0 = vertices[0,0,:].reshape(2,)
        v1, k1 = find_next_vertice(vertices,0)
        v2, k2 = find_next_vertice(vertices,k1)
        v3, k3 = find_next_vertice(vertices,k2)
        v20 = v2[0].astype(int)
        v21 = v2[1].astype(int)
        x = vertices.min(0).reshape(2,)
        y = vertices.max(0).reshape(2,)
        t= ((x+y)/2).astype(int)
        if (v0[0]-v1[0]) == 0:
            angle = 0

        else:
            slope =  ((v0[1]-v1[1])/(v0[0]-v1[0]))
            angle  = np.arctan(slope)
        result, M = rotate_image(roi, angle, (int(t[0]),int(t[1])))

        if save:
            cv2.imwrite(PATH_DATA+"/tiles/TILE_{}.png".format(i), result[t[1]-70:t[1]+70,t[0]-70:t[0]+70,:])
        tiles.append(result[t[1]-64:t[1]+64,t[0]-64:t[0]+64,:])
        
    if show:
        for tile in tiles:
            plt.figure(figsize=(1,1))
            plt.imshow(tile)
            plt.show()
    
    return tiles
````

````python
import cv2
import numpy as np
from skimage.feature import local_binary_pattern

def extract_color_histogram(image):
    b_channel, g_channel, r_channel = cv2.split(image)

    b_hist = cv2.calcHist([b_channel], [0], None, [256], [0, 256])
    g_hist = cv2.calcHist([g_channel], [0], None, [256], [0, 256])
    r_hist = cv2.calcHist([r_channel], [0], None, [256], [0, 256])

    hist_features = np.concatenate((b_hist, g_hist, r_hist), axis=0).flatten()

    return hist_features

def extract_gabor_features(gray_image, ksize, theta, lambd, sigma, gamma):
    gabor_filter = cv2.getGaborKernel((ksize, ksize), sigma, theta, lambd, gamma)
    
    filtered_image = cv2.filter2D(gray_image, cv2.CV_8UC3, gabor_filter)

    lbp = local_binary_pattern(filtered_image, 8, 1, method='uniform') # todo

    hist, _ = np.histogram(lbp.ravel(), bins=np.arange(0, 10), range=(0, 10))

    hist = hist.astype("float")
    hist /= (hist.sum() + 1e-7)

    return hist
````

````python
from sklearn.preprocessing import normalize
from sklearn.decomposition import PCA

def get_features(puzzle_pieces):
    ksizes = [7]
    thetas = np.linspace(0, np.pi, 2, endpoint=False)
    lambdas = range(5, 105, 50)
    sigmas = range(10, 50, 20)
    gammas = [0.5]
    
    all_features = []

    for puzzle_piece in puzzle_pieces:
        color_histogram = extract_color_histogram(puzzle_piece)
        gray_puzzle_piece = cv2.cvtColor(puzzle_piece, cv2.COLOR_BGR2GRAY)

        gabor_features = []

        for ksize in ksizes:
            for theta in thetas:
                for lambd in lambdas:
                    for sigma in sigmas:
                        for gamma in gammas:
                            features = extract_gabor_features(gray_puzzle_piece, ksize, theta, lambd, sigma, gamma)
                            gabor_features.append(features)

        gabor_features = np.concatenate(gabor_features, axis=0)


        features = np.concatenate((color_histogram, gabor_features), axis=0)
        all_features.append(features)

    
    print("Color Histogram shape:", color_histogram.shape)
    print("Gabor Features shape:", gabor_features.shape)
    
    all_features = normalize(all_features)

    pca = PCA(n_components=0.99)
    pca.fit(all_features)
    selected_features = pca.transform(all_features)

    return selected_features
````

````python
from sklearn.cluster import DBSCAN

def get_labels(X, eps=0.5):
    cluster = DBSCAN(eps=eps, min_samples=1).fit(X)
    labels= cluster.labels_
    print("labels:",labels)
    return labels
````

````python
def pca(X, labels, n_components=2):
    pca = PCA(n_components=n_components)
    pca_features = pca.fit_transform(X)

    plt.scatter(pca_features[:, 0], pca_features[:, 1], c=labels)
    plt.xlabel('Principal Component 1')
    plt.ylabel('Principal Component 2')
    plt.title('PCA Projection with K-means Labels')
    plt.show()
````

````python
def form_puzzles(labels, tiles):
    solved_puzzles_id = []
    outlier_images_id = []
    counter = Counter(labels)#.items()
    print("items:",counter.items())
    for i, n in list(counter.items()):
        if n in puzzle_sizes:
            solved_puzzles_id.append(i)
        else:
            outlier_images_id.append(i)
            
    solved_puzzles = []
    outlier_images = []
    for i in solved_puzzles_id:
        puzzle = []
        for k in range(len(tiles)):
            if labels[k] == i:
                puzzle.append(tiles[k])
        solved_puzzles.append(np.concatenate(puzzle))
    for i in outlier_images_id:
        for k in range(len(tiles)):
            if labels[k] == i:
                outlier_images.append(tiles[k])
                
    return solved_puzzles, outlier_images
````

````python
from sklearn.cluster import KMeans
from time import time
from collections import Counter

def solve_and_export_puzzles_image(image_index , folder = "train" , path = "data_project"  , group_id = "00", show_tiles=False, show_pca=False):
    """
    Wrapper funciton to load image and save solution
            
    Parameters
    ----------
    image:
        index number of the dataset

    Returns
    """
    # open the image
    image_loaded = load_input_image(image_index , folder = folder , path = path)
    
    tiles = segment(image_loaded, save=False, show=show_tiles)
    feats = get_features(tiles)
    labels = get_labels(feats)
    solved_puzzles, outlier_images = form_puzzles(labels, tiles)

    if show_pca:pca(feats, labels)
    
    save_solution_puzzles(image_index , solved_puzzles , outlier_images , folder = folder ,group_id =group_id)    
    return image_loaded , solved_puzzles , outlier_images
````

## Gabor filters

$$ gb(x,y) = \exp \left( -\frac{1}{2} \left( \frac{x_{\theta}^2}{\sigma^2} + \frac{y_{\theta}^2}{(\Gamma\sigma)^2} \right) \right) \cos \left( \frac{2 \pi}{\lambda} x_{\theta} + \psi \right) $$
